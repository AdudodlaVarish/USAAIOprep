```python

import numpy as np

class MyPCA:
    def __init__(self, n_components):
        self.n_components = n_components

    def fit(self, X):

        m, _ = X.shape
        self.means = np.mean(X, axis = 0)
        centered = X - self.means 
        u, sigma, vt = np.linalg.svd(centered)
        self.components = vt[:self.n_components].T
        self.n_variances = sigma[:self.n_components] ** 2 / (m - 1)
        self.explained_variances = np.cumsum(sigma[:self.n_components] ** 2 / np.sum(sigma ** 2))
        return self

    def transform(self, X):

        centered = X - self.means
        return centered @ self.components

    def inverse_transform(self, Z):

        return Z @ self.components.T + self.means
```
