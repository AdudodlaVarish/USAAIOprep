```python
import numpy as np 

def knn(inputs, data, labels, k):

    distances = np.sqrt(np.sum((data - inputs)**2, axis = -1))
    indices = np.argsort(distances)[:k]
    closest_labels = labels[indices]
    unique_labels, counts = np.unique(closest_labels, return_counts=True)
    return unique_labels[np.argmax(counts)]
```

```python

data = np.array([
    [1, 1],
    [2, 2],
    [3, 3],
    [8, 8],
    [9, 9],
    [10, 10]
])

labels = np.array([0, 0, 0, 1, 1, 1])

print(knn(np.array([2, 3]), data, labels, 3))   # Expected: 0
print(knn(np.array([9, 8]), data, labels, 3))   # Expected: 1
print(knn(np.array([1, 2]), data, labels, 1))   # Expected: 0
print(knn(np.array([10, 9]), data, labels, 1))  # Expected: 1
```
