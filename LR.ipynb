```python
import numpy as np

def LinearRegression(X, y, lr, steps):

    n, d = X.shape

    w = np.zeros(d)
    b = 0

    for _ in range(steps):

        preds = X @ w + b
        errors = preds - y
        w_gradient = (1 / n) * X.T @ errors
        b_gradient = (1/n) * np.sum(errors)
        w = w - lr * w_gradient 
        b = b - lr * b_gradient

    return w, b


```
