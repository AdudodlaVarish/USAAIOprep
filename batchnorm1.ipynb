```python

import numpy as np

def batch_norm_forward(
    X, gamma, beta,
    running_mean, running_var,
    training=True,
    momentum=0.1,
    eps=1e-5
):

    if training:

        mean = np.mean(X, axis = 0)
        variance = np.mean((X - mean) ** 2, axis = 0)

        normalized_X = (X - mean) / (np.sqrt(variance + eps))

        y = gamma * normalized_X + beta

        new_running_mean = (1 - momentum) * running_mean + momentum * mean
        new_running_var = (1 - momentum) * running_var + momentum * variance

        pass

    else:
        normalized_X = (X - running_mean) / np.sqrt(running_var + eps)

        y = gamma * normalized_X + beta
        new_running_mean = running_mean
        new_running_var = running_var

        pass

    return y, new_running_mean, new_running_var
```
