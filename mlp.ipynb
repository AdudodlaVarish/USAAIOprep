```python
import torch
import torch.nn as nn 
import torch.nn.functional as F 

class MLP(nn.Module):

    def __init__(self, embed_dim, multiplier):
        super().__init__()

        self.up = nn.Linear(embed_dim, embed_dim * multiplier, bias = False)
        self.gate = nn.Linear(embed_dim, embed_dim * multiplier, bias = False)
        self.down = nn.Linear(embed_dim * multiplier, embed_dim, bias = False)

    def forward(self, x):
        up = self.up(x)
        gate = F.silu(self.gate(x))
        return self.down(up * gate)
```
